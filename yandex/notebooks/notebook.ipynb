```python
def bin_search_left(array, x):
    result = -1
    left_idx, right_idx = 0, len(array) - 1
    while right_idx >= left_idx:
        middle_idx = (left_idx + right_idx) // 2
        middle_item = array[middle_idx]
        if middle_item == x:
            result = middle_idx
            right_idx = middle_idx - 1
        elif middle_item > x:
            right_idx = middle_idx - 1
        else:
            left_idx = middle_idx + 1
    return result 


def bin_search_right(array, x):
    result = -1
    left_idx, right_idx = 0, len(array) - 1
    while right_idx >= left_idx:
        middle_idx = (left_idx + right_idx) // 2
        middle_item = array[middle_idx]
        if middle_item == x:
            result = middle_idx
            left_idx = middle_idx + 1
        elif middle_item > x:
            right_idx = middle_idx - 1
        else:
            left_idx = middle_idx + 1
    return result

def count_freq(array, x):
    left_idx = bin_search_left(array, x)
    right_idx = bin_search_right(array, x)
    if left_idx == -1:
        return 0
    elif left_idx == right_idx:
        return 1 
    return right_idx - left_idx + 1

count_freq([1, 2, 2, 2, 3, 3, 4, 4], 4)
```

```python
def get_coverage_num(array, l):
    ans = 1 
    cur_idx = 1
    cur_len = 0
    while cur_idx < len(array):
        if cur_len + (array[cur_idx] -  array[cur_idx-1]) > l:
            ans += 1
            cur_len = 0
        else:
            cur_len += array[cur_idx] -  array[cur_idx-1]
        cur_idx += 1
    return ans
        
        


def min_coverage_len(array, k):
    left = 0
    right = array[-1] - array[0]
    ans = 0
    while right >= left:
        middle = (left + right) // 2
        if get_coverage_num(array, middle) >= k:
            ans = middle
            right = middle
        else:
            left = middle + 1
    return ans
        
min_coverage_len([2, 3, 9, 11, 20], 3)
```

```python
def get_coverage_num(array, L):
    count = 1
    current_end = array[0] + L
    for i in range(1, len(array)):
        if array[i] > current_end:
            count += 1
            current_end = array[i] + L
    return count

def min_coverage_len(array, k):
    left = 0
    right = array[-1] - array[0]
    ans = right
    while left <= right:
        mid = (left + right) // 2
        if get_coverage_num(array, mid) <= k:
            ans = mid
            right = mid - 1
        else:
            left = mid + 1
    return ans

min_coverage_len([1, 2, 1, 2], 2)
```
