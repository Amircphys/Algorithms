### Задача 1

Дан массив целых чисел arr и целое число $X$. 
Определите, существует ли в массиве такой непрерывный подмассив, что сумма его элементов равна $X$.

```python
def subarray_sum_1(non_negative_arr, target):
    right, current_sum = 0, 0
    for left in range(len(non_negative_arr)):
        # пересчитываем сумму
        if left > 0:
            current_sum -= non_negative_arr[left - 1]

        # при необходимости двигаем правую границу
        while right < len(non_negative_arr) and current_sum < target:
            current_sum += non_negative_arr[right]
            right += 1

        if current_sum == target:
            return True

    return False


def subarray_sum(arr, target):
    prefix_sums = set([0])
    temp_sum = 0
    for i in range(len(arr)):
        temp_sum += arr[i]
        if temp_sum - target in prefix_sums:
            return True
        prefix_sums.add(temp_sum)
    return False

assert subarray_sum([2, 5, 5, 19], 10) == True
assert subarray_sum([2, 5, 5, 19], 12) == True
assert subarray_sum([2, 5, 5, 19], 13) == False
```

сохранить исходный порядок

```python
def move_item(arr, item):
    slow_idx = 0
    for fast_idx in range(len(arr)):
        if arr[fast_idx] != item:
            arr[fast_idx], arr[slow_idx] = arr[slow_idx], arr[fast_idx]
            slow_idx += 1
    return arr 


assert move_item([1, 2, 3, 4], 3) == [1, 2, 4, 3]
assert move_item([3, 2, 1, 4], 3) == [2, 1, 4, 3]
assert move_item([1, 2, 3, 4], 5) == [1, 2, 3, 4]
assert move_item([1, 2, 3, 3, 4], 3) == [1, 2, 4, 3, 3]
assert move_item([1, 3, 2, 2, 3, 4], 3) == [1, 2, 2, 4, 3, 3]
assert move_item([3, 3, 2, 2, 3, 4], 3) == [2, 2, 4, 3, 3, 3]
assert move_item([3, 3, 2, 2, 3, 4], 1) == [3, 3, 2, 2, 3, 4]
```

### Задача 3
Даны два отсортированных массива, необходимо получить третий отсортированный массив.

```python
def merge_arrays(arr_1, arr_2):
    idx_1, idx_2, idx_3 = 0, 0, 0
    result = [None for _ in range(len(arr_1)+len(arr_2))]
    while idx_1 < len(arr_1) and idx_2 < len(arr_2):
        if arr_1[idx_1] < arr_2[idx_2]:
            result[idx_3] = arr_1[idx_1]
            idx_1 += 1
        else:
            result[idx_3] = arr_2[idx_2]
            idx_2 += 1
            
        idx_3 += 1
    while idx_1 < len(arr_1):
        result[idx_3] = arr_1[idx_1]
        idx_1 += 1
        idx_3 += 1
        
    while idx_2 < len(arr_2):
        result[idx_3] = arr_2[idx_2]
        idx_2 += 1
        idx_3 += 1
        
    return result             


assert merge_arrays([], []) == []
assert merge_arrays([1], []) == [1]
assert merge_arrays([], [1]) == [1]
assert merge_arrays([1], [1]) == [1, 1]
assert merge_arrays([1], [2]) == [1, 2]
assert merge_arrays([1, 2, 3], [2]) == [1, 2, 2, 3]
assert merge_arrays([2], [1]) == [1, 2]
assert merge_arrays([2, 5], [1]) == [1, 2, 5]
assert merge_arrays([2, 5, 5], [1, 3, 4]) == [1, 2, 3, 4, 5, 5]
```

### Задача 4

Дан остортированный массив целых чисел и число target. Необходимо вернуть индексы двух значений, которые в сумме дают target. При этом каждый элемент может использоваться только один раз. 

```python
def two_sum(arr, target):
    left_idx, right_idx = 0, len(arr)-1
    while left_idx < right_idx:
        temp_sum = arr[left_idx] + arr[right_idx]
        if temp_sum == target:
            return (left_idx, right_idx)
        elif temp_sum < target:
            left_idx += 1
        else:
            right_idx -= 1 


assert two_sum([1, 2, 5], 7) == (1, 2)
assert two_sum([1, 2, 5, 6], 3) == (0, 1)
assert two_sum([1, 2, 5, 6], 6) == (0, 2)
```

```python
"""
[1, 1, 2] -> [1, 2]
[1, 2, 2, 3] -> [1, 2, 3]
[1, 2, 3, 3] -> [1, 2, 3]
[1, 2, 3] -> [1, 2, 3]
"""

class ListNode(object):
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
        
class Solution(object):
    def deleteDuplicates(self, head):
        """
        :type head: Optional[ListNode]
        :rtype: Optional[ListNode]
        """
        temp_node = head
        next_node = head
        while temp_node:
            while next_node and next_node.val == temp_node.val:
                next_node = next_node.next
            temp_node.next = next_node
            temp_node = next_node
            
        return head
    
    
class Solution_2(object):
    def deleteDuplicates(self, head):
        """
        :type head: Optional[ListNode]
        :rtype: Optional[ListNode]
        """
        cur_node = head
        while cur_node and cur_node.next:
            while cur_node.next and cur_node.val == cur_node.next.val:
                cur_node.next = cur_node.next.next
            cur_node = cur_node.next
        return head 
```

### Задача 7

```python

"""
[1, 1, 2] -> [2]
[1, 2, 2, 3] -> [1, 3]
[1, 2, 3, 3] -> [1, 2]
[1, 2, 3] -> [1, 2, 3]
"""
class ListNode(object):
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
        
        
class Solution(object):
    def deleteDuplicates(self, head):
        """
        :type head: Optional[ListNode]
        :rtype: Optional[ListNode]
        """
        current = head
        dummy = ListNode(val=None, next=head)
        prev = dummy 
        while current:
            if current.next and current.next.val == current.val:
                while current.next and current.next.val == current.val:
                    current = current.next
                prev.next = current.next 
            else:
                prev = current
            current = current.next
                
        return dummy.next
                
```

### Задача 8

Дан массив целых чисел nums, отсортированный по неубывающему порядку. Удалить дубликаты на месте так, чтобы каждый уникальный элемент встречался только один раз. Относительный порядок элементов должен оставаться неизменным.

Считайте количество уникальных элементов в массиве nums равным k. После удаления дубликатов верните количество уникальных элементов k.

Первые k элементов массива nums должны содержать уникальные числа в отсортированном порядке. Остальные элементы после индекса k - 1 можно игнорировать.

```python
"""
Input: nums = [0,0,1,1,1,2,2,3,3,4]
Output: 5, nums = [0,1,2,3,4,_,_,_,_,_]

Input: nums = [1,1,2]
Output: 2, nums = [1,2,_]
"""




class Solution(object):
    def removeDuplicates(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        slow_idx = 0
        for fast_idx in range(len(nums)):
            if nums[fast_idx] != nums[slow_idx]:
                nums[slow_idx+1] = nums[fast_idx]
                slow_idx += 1
        for i in range(slow_idx+1, len(nums)):
            nums[i] = "_"
            
        return nums
    
Solution().removeDuplicates([1,1,2])
```

### Задача 9

Даны два целочисленных массива nums1 и nums2. 

Верните массив их пересечения. Каждый элемент в результате должен быть уникальным, и результаты можно возвращать в любом порядке.

```python
""" 
Example 1:

Input: nums1 = [1,2,2,1], nums2 = [2,2]
Output: [2]
Example 2:

Input: nums1 = [4,9,5], nums2 = [9,4,9,8,4]
Output: [9,4]
Explanation: [4,9] is also accepted.

"""


class Solution(object):
    def intersection(self, nums1, nums2):
        nums1.sort()
        nums2.sort()
        
        i, j = 0, 0
        result = []
        
        while i < len(nums1) and j < len(nums2):
            if nums1[i] < nums2[j]:
                i += 1
            elif nums1[i] > nums2[j]:
                j += 1
            else:
                # Добавляем элемент, только если он не равен предыдущему добавленному
                if not result or result[-1] != nums1[i]:
                    result.append(nums1[i])
                i += 1
                j += 1
        
        return result
    
Solution().intersection(nums1 = [4,9,5], nums2 = [9,4,9,8,4])
```

### Задача 10
Даны две строки: «needle» и «haystack», вернуть индекс первого вхождения «needle» в «haystack» или -1, если «needle» не является частью «haystack».

```python
class Solution(object):
    def strStr(self, haystack, needle):
        """
        :type haystack: str
        :type needle: str
        :rtype: int
        """
        if not needle:  # если needle пустая строка
            return 0
            
        for i in range(len(haystack) - len(needle) + 1):
            found = True  # сбрасываем для каждого i
            for j in range(len(needle)):
                if needle[j] != haystack[i + j]:
                    found = False
                    break  # прерываем при первом несовпадении
            if found:
                return i
        return -1

Solution().strStr(haystack = "sabbutsad", needle = "ts")
```

###  Задача 11

Фраза является палиндромом, если после преобразования всех заглавных букв в строчные и удаления всех символов, кроме букв и цифр, она читается одинаково слева направо и справа налево. К буквенно-цифровым символам относятся буквы и цифры.

Для строки s вернуть true, если она является палиндромом, или false в противном случае.

```python
"""
Input: s = "A man, a plan, a canal: Panama"
Output: true
Explanation: "amanaplanacanalpanama" is a palindrome.

Input: s = "race a car"
Output: false
Explanation: "raceacar" is not a palindrome.

"""


class Solution(object):
    def isPalindrome(self, s):
        """
        :type s: str
        :rtype: bool
        """
        symbols = 'abcdefghijklmnopqrstuvwxyz1234567890'
        left_idx, right_idx = 0, len(s)-1
        s = s.lower()
        while left_idx < right_idx:
            while (s[left_idx] not in symbols) and left_idx < right_idx:
                left_idx += 1
            while (s[right_idx] not in symbols) and right_idx > left_idx:
                right_idx -= 1
            if s[left_idx] != s[right_idx]:
                return False
            left_idx += 1
            right_idx -= 1
        return True
    
    

Solution().isPalindrome("rac   a car")   
```

```python

class ListNode(object):
    def __init__(self, x):
        self.val = x
        self.next = None

class Solution(object):
    def getIntersectionNode(self, headA, headB):
        """
        :type head1, head1: ListNode
        :rtype: ListNode
        """
        first = headA
        second = headB
        while first != second:
            first = first.next if first else headB
            second = second.next if second else headA

        return first
            
```

### Задача 15

Напишите функцию, которая переворачивает строку. Входная строка представлена ​​массивом символов s.

Для этого необходимо изменить входной массив на месте, используя O(1) дополнительной памяти.

```python
"""
Input: s = ["h","e","l","l","o"]
Output: ["o","l","l","e","h"]
"""

class Solution(object):
    def reverseString(self, s):
        """
        :type s: List[str]
        :rtype: None Do not return anything, modify s in-place instead.
        """
        left_idx, right_idx = 0, len(s)-1
        while left_idx < right_idx:
            s[left_idx], s[right_idx] = s[right_idx], s[left_idx]
            left_idx += 1
            right_idx -= 1
        return s
    
Solution().reverseString(["h","e","l","l","o"])
```

### Задача 16

Дана строка s, перевернуть только все гласные в ней и вернуть её.

Гласные — это «a», «e», «i», «o» и «u», и они могут встречаться как в нижнем, так и в верхнем регистре более одного раза.

```python
""" 
Input: s = "IceCreAm"

Output: "AceCreIm"

"""
class Solution(object):
    def reverseVowels(self, s):
        v = 'aeiou'
        vowels = set(list(v) + list(v.upper()))
        left_idx, right_idx = 0, len(s) - 1
        s_list = list(s)  # создаем список из строки

        while left_idx < right_idx:
            # Если левый символ не гласная - двигаем левый указатель
            if s_list[left_idx] not in vowels:
                left_idx += 1
            # Если правый символ не гласная - двигаем правый указатель
            elif s_list[right_idx] not in vowels:
                right_idx -= 1
            else:
                # Оба символа - гласные, меняем местами
                s_list[left_idx], s_list[right_idx] = s_list[right_idx], s_list[left_idx]
                left_idx += 1
                right_idx -= 1

        return "".join(s_list)
    
    
Solution().reverseVowels("IceCreAm")
```

```python
""" 
Input: nums1 = [1,2,2,1], nums2 = [2,2]
Output: [2,2]

Input: nums1 = [4,9,5], nums2 = [9,4,9,8,4]
Output: [4,9]
Explanation: [9,4] is also accepted.
"""

class Solution(object):
    def intersect(self, nums1, nums2):
        """
        :type nums1: List[int]
        :type nums2: List[int]
        :rtype: List[int]
        """
        nums1.sort()
        nums2.sort() 
        
        result = []
        i1, i2 = 0, 0 
        while i1 < len(nums1) and i2 < len(nums2):
            if nums1[i1] == nums2[i2]:
                result.append(nums1[i1])
                i1 += 1
                i2 += 1
            elif nums1[i1] < nums2[i2]:
                i1 += 1 
            else:
                i2 += 1
        return result
    
Solution().intersect(nums1 = [1,2,2,1], nums2 = [2,2])
        
```

### Задача 18

Даны две строки s и t. Возвращает true, если s является подпоследовательностью t, или false в противном случае.

Подпоследовательность строки — это новая строка, образованная из исходной строки путём удаления некоторых (возможно, ни одного) символов без нарушения относительного расположения оставшихся символов. (т.е. «ace» является подпоследовательностью «abcde», а «aec» — нет).

```python
""" 
Example 1:
Input: s = "abc", t = "ahbgdc"
Output: true


Example 2:
Input: s = "axc", t = "ahbgdc"
Output: false

"""

class Solution(object):
    def isSubsequence(self, s, t):
        """
        :type s: str
        :type t: str
        :rtype: bool
        """
        if len(s) == 0:
            return True
        idx_s, idx_t = 0, 0 
        while idx_t < len(t):
            if t[idx_t] == s[idx_s]:
                idx_s += 1 
                if idx_s == len(s):
                    return True
            idx_t += 1 
        return False
    
Solution().isSubsequence(s = "abc", t = "ahbgd")
```

### Задача 19

Предположим, вы отличный родитель и хотите побаловать своих детей печеньем. Но каждому ребёнку следует давать не более одного печенья.

У каждого ребёнка i есть фактор жадности g[i], который представляет собой минимальный размер печенья, которым ребёнок будет доволен; а у каждого печенья j есть размер s[j]. Если s[j] >= g[i], мы можем назначить печенье j потомку i, и он будет доволен. Ваша цель — максимизировать количество довольных потомков и вывести максимальное количество.

```python
""" 
g = [10,9,8,7]
s = [5,6,7,8]
2
"""


class Solution(object):
    def findContentChildren(self, g, s):
        """
        :type g: List[int]
        :type s: List[int]
        :rtype: int
        """
        g.sort()
        s.sort()
        res = 0
        for i in range(min(len(g), len(s))):
            if s[i] >= g[i]:
                res += 1
        return res
```
